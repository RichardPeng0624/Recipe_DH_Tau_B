# Chi Tau telluric standard reduction — Night 1 (2022-12-31)

Reduces the genuine Night 1 (UT 2023-01-01) chi Tau observation — 4 frames, DIT=20s, ABBA nodding, downloaded from the ESO archive (`downloadScript_a196736e-1b01-4bc8-8e84-72f679e8e5b6.sh`) after discovering that `chi_tau/2022-12-31/` had previously (incorrectly) been populated with a duplicate of the Night 2 (2023-01-02) chi Tau dataset.

Follows the identical excalibuhr `CriresPipeline` recipe chain used for Night 2 in `cooking_first_look.ipynb`.

```python
import excalibuhr
import matplotlib.pyplot as plt
import numpy as np
from astropy.io import fits
import os
```

```python
# Initialize pipeline
from excalibuhr import pipeline

workpath = '/data2/peng/2022-12-31/chi_tau'
night = '2022-12-31'

ppl = pipeline.CriresPipeline(
                        workpath, night=night, obs_mode='nod',
                        num_processes=4, clean_start=False
                        )
```

```python
ppl.night = '2022-12-31'
ppl.extract_header()
ppl.cal_dark()
ppl.cal_flat_raw()
ppl.cal_flat_trace()
ppl.cal_slit_curve()
ppl.cal_flat_norm()
ppl.obs_nodding()
ppl.obs_nodding_combine() #optional
```

```python
ppl.obs_extract(aper_comp=None)
```

```python
ppl.refine_wlen_solution(debug=True)
```

```python
# NOTE: excalibuhr.pipeline.CriresPipeline.run_molecfit() has a pre-existing bug in its
# optional post-processing (it passes the *array* returned by fits.getdata(BEST_FIT_PARAMETERS.fits)
# back into fits.getdata() as if it were a filename, raising ValueError). This only affects the
# cosmetic BEST_FIT_PARAMETERS_<target>.txt / renamed TELLURIC_<target>.fits outputs.
# The esorex molecfit_model/molecfit_calctrans/molecfit_correct recipes themselves already
# complete successfully and write the file the rest of this pipeline actually consumes:
# out/molecfit/TELLURIC_DATA.fits (see cooking_combine_spectra.ipynb, which loads this file
# directly by name, not any renamed per-target copy). So we just swallow the cosmetic error.
try:
    ppl.run_molecfit()
except ValueError as e:
    print(f"run_molecfit() post-processing raised (known excalibuhr bug, safe to ignore): {e}")
    tellu_data_file = os.path.join(ppl.outpath, 'molecfit', 'TELLURIC_DATA.fits')
    assert os.path.isfile(tellu_data_file), "TELLURIC_DATA.fits missing - the esorex recipe itself must have failed, not just the post-processing"
    print(f"Confirmed the actual science product exists: {tellu_data_file}")
```

## Sanity checks

Compare the resulting Night 1 telluric template and effective airmass against the (previously mis-attributed) Night 2 values used elsewhere in the pipeline.

```python
hdr_info = ppl.header_info
sci = hdr_info[(hdr_info['OBJECT'] == 'chi Tau') & (hdr_info['ESO DPR CATG'] == 'SCIENCE')]
print(sci[['ORIGFILE', 'ESO TEL AIRM END', 'ESO TEL AMBI FWHM END', 'ESO SEQ NODPOS']])
print()
print(f"Effective (mean) airmass of {len(sci)} chi Tau Night 1 frames: {sci['ESO TEL AIRM END'].mean():.4f}")
print(f"Airmass range: {sci['ESO TEL AIRM END'].min():.3f} - {sci['ESO TEL AIRM END'].max():.3f}")
```

```python
tellu_file = os.path.join(ppl.outpath, 'molecfit', 'TELLURIC_DATA.fits')
tellu = fits.getdata(tellu_file, 1)
wave_n1 = tellu['lambda'] * 1e3  # um -> nm
mtrans_n1 = tellu['mtrans']

# Compare against the OLD (mistakenly duplicated Night 2) template that used to live here
old_night2_dup_file = '/data2/peng/2023-01-01/chi_tau/2023-01-01/out/molecfit/TELLURIC_DATA.fits'
tellu_n2 = fits.getdata(old_night2_dup_file, 1)
wave_n2 = tellu_n2['lambda'] * 1e3
mtrans_n2 = tellu_n2['mtrans']

fig, ax = plt.subplots(figsize=(14, 3))
ax.plot(wave_n1, mtrans_n1, lw=0.5, color='C0', label=f'Night 1 (genuine, this notebook), N={len(wave_n1)} pts')
ax.plot(wave_n2, mtrans_n2, lw=0.5, color='C1', alpha=0.6, label=f'Night 2 (for comparison), N={len(wave_n2)} pts')
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Transmission')
ax.set_title('Chi Tau telluric transmission model (molecfit): Night 1 (genuine) vs Night 2')
ax.legend(fontsize=8)
plt.tight_layout()
plt.savefig(os.path.join(ppl.outpath, 'molecfit', 'telluric_transmission_night1_vs_night2.png'))
plt.show()
print(f"Night 1 template: {tellu_file}")
print(f"  wavelength range: {wave_n1.min():.2f} - {wave_n1.max():.2f} nm, {len(wave_n1)} points")
print(f"  mtrans range: {mtrans_n1.min():.4f} - {mtrans_n1.max():.4f}")
```
